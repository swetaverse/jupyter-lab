```python
import matplotlib.pyplot as plt
import numpy as np

# Wavelengths
wavelengths = np.linspace(400, 700, 31)

# OD values
OD_values = [
    0.52, 0.50, 0.49, 0.48, 0.475, 0.47, 0.472, 0.48, 0.49, 0.505, 0.52,
    0.54, 0.56, 0.58, 0.59, 0.595, 0.593, 0.585, 0.57, 0.55, 0.53,
    0.51, 0.49, 0.47, 0.45, 0.43, 0.41, 0.39, 0.37, 0.35, 0.33
]

# Find peak value and corresponding wavelength
peak_index = np.argmax(OD_values)
peak_wavelength = wavelengths[peak_index]
peak_value = OD_values[peak_index]

# Plot
plt.figure(figsize=(10, 6))
plt.plot(wavelengths, OD_values, 'r.', markersize=6, label='Spectrum D5')

# Draw vertical and horizontal dotted lines from the peak
plt.axvline(peak_wavelength, ymin=0, ymax=(peak_value - 0.32) / (0.72 - 0.32),
            color='red', linestyle='dotted', linewidth=1)
plt.axhline(peak_value, xmin=(peak_wavelength - 400) / (700 - 400), xmax=1,
            color='red', linestyle='dotted', linewidth=1)

# Set limits
plt.xlim(400, 700)
plt.ylim(0.32, 0.72)

# Labels and layout
plt.xlabel("Wavelength (nm)")
plt.ylabel("Absorbance / OD")
plt.title("Spectrum - D5 (with Peak Highlighted)")
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```
